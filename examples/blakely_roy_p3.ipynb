### Blakely-Roy inequality

This note computes 
$$
  \min_G t(P_3, G) - t(e, G)^3
$$
which is also known as Blakely-Roy inequality, and is special case of Sidorenko's conjecture.

It is well known that the minimum value is zero, however, also known that flag algebra based calculus can not give the answer of this inequality: <https://arxiv.org/abs/1812.08820>.

```python
import flag_algebra.flag_algebra as fa
import flag_algebra.build_problem as bp
import networkx as nx
import cvxpy as cp
```

```python
P3 = nx.from_edgelist([(0, 1), (1, 2), (2, 3)]) # Use t(P3, G).
e3 = nx.from_edgelist([(0, 1), (2, 3), (4, 5)]) # Use t(e U e U e, G)  = t(e, G)^3.
```

```python
print("Building problem...")
problem, variable_dict = bp.build_problem(
  objectives = [('sub', P3, 1/3), ('sub', e3, -1/15)], # t(P3, G) = p(P3, G) / 3, t(e U e U e, G) = p(e U e U e, G) / 15
  constraints = [],
  sdp_configs = [(4, 1), (5, 3), (6, 5)],
  lowerbound = True
)
print("Solving problem...")
problem.solve(solver=cp.SCS, verbose=True)

print("Optimal value:", problem.value)
```
