### Goodman's Bound

This notes computes
$$
  \min_G t(\triangle, G) - 2 t(e, G)^2 + t(e, G)
$$
which is known as the Goodman's bound.

It is known that the minimum is zero (i.e., the inequality $t(\triangle, G) \ge 2 t(e, G)^2 - t(e, G)$ holds), and is provable via flag algebra based calculus.

```python
import flag_algebra.flag_algebra as fa
import flag_algebra.build_problem as bp
import networkx as nx
import cvxpy as cp
```

```python
k3 = nx.from_edgelist([(0, 1), (1, 2), (2, 0)]) # Use t(K3, G).
e2 = nx.from_edgelist([(0, 1), (2, 3)])         # Use t(e U e, G) = t(e, G)^2.
e1 = nx.from_edgelist([(0, 1)])                 # Use t(e, G).
```

```python
print("Building problem...")

atlas = fa.get_graph_atlas(7)
sdp_41 = fa.compute_grouped_averaged_flag_product_coefficients(atlas, 4, 1, verbose=True)
sdp_53 = fa.compute_grouped_averaged_flag_product_coefficients(atlas, 5, 3, verbose=True)
sdp_65 = fa.compute_grouped_averaged_flag_product_coefficients(atlas, 6, 5, verbose=True)

vertex_differential = fa.vertex_differential([('sub', k3, 1), ('sub', e2, -2), ('sub', e1, 1)], atlas)
edge_differential = fa.edge_differential([('sub', k3, 1), ('sub', e2, -2), ('sub', e1, 1)], atlas)

problem, variable_dict = bp.build_problem(
  objectives = [('sub', k3, 1), ('sub', e2, -2), ('sub', e1, 1)],
  constraints=[],
  sdp_configs=[(4, 1, sdp_41), (5, 3, sdp_53), (6, 5, sdp_65)],
  lowerbound = True
)
print("Solving problem...")
problem.solve(solver=cp.SCS, verbose=True)

print("Optimal value:", problem.value)
```

```python

problem, variable_dict = bp.build_problem(
  objectives = [('sub', k3, 1), ('sub', e2, -2), ('sub', e1, 1)],
  constraints=[],
  sdp_configs=[(4, 1, sdp_41), (5, 3, sdp_53), (6, 5, sdp_65)],
  lowerbound = True,
  use_edge_differential=edge_differential,
)
print("Solving problem...")
problem.solve(solver=cp.SCS, verbose=True)

print("Optimal value:", problem.value)
```
